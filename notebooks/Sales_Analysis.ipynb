# Sales Data Analysis

**Summer Internship 2026 — Data Analysis Project**

This project analyzes sales performance, profitability, products, regions, and monthly trends using Python, Pandas, Matplotlib, and Seaborn.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme()
df = pd.read_csv('../data/sales_data.csv')
df['Date'] = pd.to_datetime(df['Date'])
df.head()
```

## 1. Data Quality Check

```python
print('Rows:', len(df))
print('Columns:', len(df.columns))
print('\nMissing values:\n', df.isnull().sum())
print('\nDuplicate rows:', df.duplicated().sum())
```

## 2. Overall Performance

```python
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
total_units = df['Quantity'].sum()
print(f'Total Sales: ₹{total_sales:,.2f}')
print(f'Total Profit: ₹{total_profit:,.2f}')
print(f'Total Units Sold: {total_units:,}')
print(f'Profit Margin: {total_profit/total_sales:.2%}')
```

## 3. Product Performance

```python
product_summary = df.groupby('Product').agg(Sales=('Sales','sum'), Profit=('Profit','sum'), Quantity=('Quantity','sum')).sort_values('Sales', ascending=False)
product_summary
```

```python
product_summary['Sales'].plot(kind='bar', title='Sales by Product', ylabel='Sales (₹)', xlabel='Product', rot=35)
plt.tight_layout()
plt.savefig('../visualizations/top_products.png', dpi=150)
plt.show()
```

## 4. Regional Performance

```python
region_summary = df.groupby('Region').agg(Sales=('Sales','sum'), Profit=('Profit','sum'), Quantity=('Quantity','sum')).sort_values('Sales', ascending=False)
region_summary
```

```python
region_summary['Sales'].plot(kind='bar', title='Sales by Region', ylabel='Sales (₹)', xlabel='Region')
plt.tight_layout()
plt.savefig('../visualizations/regional_sales.png', dpi=150)
plt.show()
```

## 5. Monthly Sales Trend

```python
monthly = df.groupby(df['Date'].dt.to_period('M')).agg(Sales=('Sales','sum'), Profit=('Profit','sum'))
monthly.index = monthly.index.astype(str)
monthly
```

```python
monthly['Sales'].plot(kind='line', marker='o', title='Monthly Sales Trend', ylabel='Sales (₹)', xlabel='Month')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('../visualizations/monthly_sales.png', dpi=150)
plt.show()
```

## 6. Category Profitability

```python
category = df.groupby('Category').agg(Sales=('Sales','sum'), Profit=('Profit','sum')).sort_values('Profit', ascending=False)
category
```

```python
category['Profit'].plot(kind='bar', title='Profit by Category', ylabel='Profit (₹)', xlabel='Category', rot=0)
plt.tight_layout()
plt.savefig('../visualizations/category_profit.png', dpi=150)
plt.show()
```

## 7. Key Findings

- Identify the highest-revenue product from `product_summary`.
- Compare regions using `region_summary` to identify the strongest market.
- Use `monthly` to identify peak and low sales months.
- Compare category profit to understand which category contributes most to profitability.

**Conclusion:** The analysis demonstrates a complete data-analysis workflow: data loading, quality checking, KPI calculation, aggregation, visualization, and business insights.
